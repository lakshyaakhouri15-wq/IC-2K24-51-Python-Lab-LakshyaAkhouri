# Lab 4 — Section C: Count Pairs With a Given Sum

**Task:** Count pairs of distinct positions whose values add up to a target. A pair is counted once, with the earlier position first.

Two strategies are compared: checking every pair and using a dictionary to remember values already encountered.

```python
import random
from time import perf_counter


def count_pairs_nested(values, target):
    """Count matching index pairs by examining every possible pair."""
    matches = 0

    for first in range(len(values)):
        for second in range(first + 1, len(values)):
            if values[first] + values[second] == target:
                matches += 1

    return matches


def count_pairs_frequency(values, target):
    """Count matching pairs using frequencies of previously seen values."""
    matches = 0
    earlier_counts = {}

    for current in values:
        complement = target - current
        matches += earlier_counts.get(complement, 0)
        earlier_counts[current] = earlier_counts.get(current, 0) + 1

    return matches
```

```python
# Use a fixed seed so that repeated runs generate the same test data.
random.seed(42)
small_values = [random.randint(1, 20_000) for _ in range(5_000)]
large_values = [random.randint(1, 20_000) for _ in range(50_000)]
target_sum = 20_000

# The nested-loop method is measured on both sizes.
for sample in (small_values, large_values):
    started = perf_counter()
    pair_total = count_pairs_nested(sample, target_sum)
    elapsed = perf_counter() - started
    print(f"Nested loops | N={len(sample):,} | pairs={pair_total:,} | {elapsed:.4f}s")

# The dictionary method is measured on the larger data set.
started = perf_counter()
pair_total = count_pairs_frequency(large_values, target_sum)
elapsed = perf_counter() - started
print(f"Dictionary   | N={len(large_values):,} | pairs={pair_total:,} | {elapsed:.4f}s")

# Verify both methods agree on a smaller input.
print("Methods agree on sample:",
      count_pairs_nested(small_values, target_sum) ==
      count_pairs_frequency(small_values, target_sum))
```

## Observations

**Q7. What happened when N increased by 10 times?**

The nested-loop method became much slower because it considers all pairs of positions. The number of pair checks grows approximately with the square of the input size.

**Q8. Compare the two methods for N = 50,000.**

The dictionary method should be substantially faster than the nested-loop method. Run the timing cell on your own computer for the actual measurements; execution times vary by system.

**Q9. What made the second method faster?**

The dictionary stores the frequency of each value already visited. For a new value, the program checks how many earlier values would complete the target sum. This avoids scanning every possible partner.
