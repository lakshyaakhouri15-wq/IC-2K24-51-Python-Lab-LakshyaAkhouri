### 1. What is the main lesson from all three problems?

A suitable algorithm can solve the same problem with fewer operations and less execution time.

### 2. Would you change your first answer in Section A?

Yes. Short code is not necessarily fast code. The algorithm and the amount of repeated work have a major effect on performance.

### 3. Which technique would you use again, and why?

I would use dictionaries for hash-based lookup. They make it possible to find previously seen values quickly and can replace repeated comparisons in searching and counting tasks.
